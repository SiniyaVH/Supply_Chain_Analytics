#### Duplicate Analysis

```python
print("Duplicate Rows:", df.duplicated().sum())

print(
    "Order Item IDs:",
    df['Order Item Id'].nunique()
)
print(
    "Order IDs:",
    df['Order Id'].nunique()
)
```

Here 'Order Item Id' contains unique values. 
Important: Multiple rows with the same Order Id do not automatically mean duplicate records. One order may contain multiple product items.

```python
duplicate_columns = []
duplicate_columns1 = []
for i in range(len(df.columns)):
    for j in range(i + 1, len(df.columns)):

        if df.iloc[:, i].equals(df.iloc[:, j]):
            duplicate_columns.append(df.columns[j])
            duplicate_columns1.append(df.columns[i])

print("Duplicate columns:", duplicate_columns)
print("Duplicated of:", duplicate_columns1)
```

```python
df.drop(columns=duplicate_columns1, inplace=True,errors='ignore')
print("Columns to be dropped:", duplicate_columns1)
print("Remaining columns:")
df.columns.tolist()
```

```python
col_drop=["Customer Password","Customer Password","Customer Zipcode","Order Zipcode",
          "Product Description","Product Image","Product Card Id","Customer City","Customer Country","Latitude","Longitude"]

existing_columns = [col for col in col_drop if col in df.columns]
missing_columns = [col for col in col_drop if col not in df.columns]

print("Columns to be deleted:", existing_columns)
print("Columns not found:", missing_columns)

df.drop(columns=col_drop, inplace=True, errors='ignore')
print("\n Remaining columns:",df.columns.tolist())
```

```python
df["Customer Name"] = (
    df["Customer Fname"].fillna('') + " " +
    df["Customer Lname"].fillna('')
).str.strip()
df.drop(columns=["Customer Fname","Customer Lname"], inplace=True, errors='ignore')
df["Customer Name"].head()
```

```python
df.info()
```

```python
date_columns = ["shipping date (DateOrders)", "order date (DateOrders)"]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')
df['Product Status'] = df['Product Status'].astype(bool)
df['Late_delivery_risk'] = df['Late_delivery_risk'].astype(bool)
```

```python
print("Shipping Date")
df["shipping date (DateOrders)"]
```

```python
print("Order date")
df["order date (DateOrders)"]
```

```python
df["Actual Shipping Days"] = (df["shipping date (DateOrders)"]-df["order date (DateOrders)"]).dt.days
df["Actual Shipping Days"].equals(df["Days for shipping (real)"])
```

```python
comparison = pd.DataFrame({
    'Calculated Days': df['Actual Shipping Days'],
    'Original Days': df['Days for shipping (real)']
})

comparison['Difference'] = (
    comparison['Calculated Days'] -
    comparison['Original Days']
)

comparison['Difference'].value_counts().sort_index()
```

```python
c = ["Days for shipping (real)"]
column_dropped = df.drop(columns = c, inplace=True, errors='ignore')
print("Columns dropped")
```

```python
df.rename(
    columns={
        'order date (DateOrders)': 'Order Date', 'shipping date (DateOrders)' : 'Shipping date'
        , 'Days for shipment (scheduled)' : 'Scheduled Shipping days', 'Delivery Status' : 'Delivery_Status',
        'Order Item Discount' : 'Discount', 'Order Customer Id' : 'Customer Id'
    },
    inplace=True
)
```

```python
df.info()
```

```python
Shipping_cancelled = (df['Delivery_Status'] == 'Shipping canceled').sum()

df.drop(df[df['Delivery_Status'] == 'Shipping canceled'].index, inplace=True)

df = df[ df['Delivery_Status'].str.strip().str.lower()!= 'shipping canceled']
print("Shipping canceled rows deleted:", Shipping_cancelled)
print("Remaining rows:", len(df))
```

```python
conditions = [
    df['Scheduled Shipping days'] > df['Actual Shipping Days'],
    df['Scheduled Shipping days'] < df['Actual Shipping Days'],
    df['Scheduled Shipping days'] == df['Actual Shipping Days']
]

status = [
    'Advanced Shipping',
    'Late Delivery',
    'Shipping On Time'
]

df['Delivery Status'] = np.select(conditions, status, default='Unknown')
df['Delivery Status']
```

```python
df['Delivery Status'].equals(df['Delivery_Status'])
```

```python
df.drop(columns = 'Delivery_Status', inplace=True, errors='ignore')
df.info()
```

```python
delivery_risk = []

for status in df['Delivery Status']:
    if status == 'Late Delivery':
        delivery_risk.append(1)
    else:
        delivery_risk.append(0)

df['Delivery Risk'] = delivery_risk
df['Delivery Risk']
```

```python
df["Delivery Risk"].equals(df['Late_delivery_risk'])
```

```python
df.drop(columns = 'Late_delivery_risk', inplace=True, errors='ignore')
df.info()
```

```python
df["Order Year"] = df['Order Date'].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.strftime('%b')
df["Order Day"] = df["Order Date"].dt.day_name()
df["Order Hour"] = df["Order Date"].dt.hour
```

```python
df["Delay"]=df["Actual Shipping Days"]-df["Scheduled Shipping days"]
df["Delay"]
```

```python
df["Is delay"]=df['Delay']>0
df["Is delay"]
```

```python
df["Profitability"]= np.where(df["Order Profit Per Order"]>0, "Profit", 
                              np.where(df["Order Profit Per Order"]<0, "Loss", "Break-even"))
df["Profitability"]
```
